```python
import array
from collections import defaultdict, deque

LAST_NAME = "Adonis"
STUDENT_ID = "TUPM-26-1828"

seed_digit = int(STUDENT_ID[-1])
id_checksum = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
vector_dim = len(LAST_NAME)

sys_config = {
    "operator": LAST_NAME,
    "auth_id": STUDENT_ID,
    "base_seed": seed_digit,
    "checksum": id_checksum,
    "vector_dim": vector_dim,
    "status": "INITIALIZED",
}

print("\n--- SYSTEM CONFIGURATION ---")
for key, value in sys_config.items():
    print(f"{key.upper()}: {value}")
```

```python
# Task 5: Mutable Sequences (Lists)
base_val = sys_config["base_seed"]
number_sequence = [base_val, base_val + 5, base_val + 15, sys_config["checksum"]]
print(f"Initial Sequence: {number_sequence}")

number_sequence.append(base_val + 20)
print(f"After Append: {number_sequence}")
```

```python
# Task 6: Sequential List Operations
new_numbers = [base_val + 5, sys_config["vector_dim"], base_val]
number_sequence.extend(new_numbers)
print(f"After Extend: {number_sequence}")

base_count = number_sequence.count(base_val)
print(f"Occurrences of {base_val}: {base_count}")

number_sequence.sort()
print(f"Sorted Sequence: {number_sequence}")
```

```python
# Task 7: Immutable Sequences (Tuples)
fixed_coordinates = (sys_config["vector_dim"], sys_config["base_seed"], 0)
print(f"Fixed Coordinates: {fixed_coordinates}")

x_val, y_val, z_val = fixed_coordinates
print(f"Unpacked Values: X: {x_val}, Y: {y_val}, Z: {z_val}")

print("Attempting modification...")
try:
    fixed_coordinates[0] = 99
except TypeError as error:
    print(f"Modification Error: {error}")
```

```python
# Task 13: Double-Ended Queues (Deque)
data_queue = deque([sys_config["base_seed"], sys_config["vector_dim"]])
print(f"Initial Deque: {data_queue}")

data_queue.append(100)
data_queue.appendleft(200)
print(f"After Appends: {data_queue}")

data_queue.popleft()
print(f"Final Deque (After Popleft): {data_queue}")
```

```python
# 1. Engineering Test Data Organizer
from collections import defaultdict

LAST_NAME = "Adonis"
SEED_NUM = 8
FAVORITE_ARTIST = "Gilbert Campo"

# Student-specific generation rules
test_sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]  # ['Adonis_T1', 'Adonis_T2', ...]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# Generate values using SEED_NUM and character positions
artist_clean = FAVORITE_ARTIST.replace(" ", "")
values = [(SEED_NUM + i + ord(char.upper()) % 10) for i, char in enumerate(artist_clean[:5])]

# 1. Generated Test Data
raw_records = [
    {"source": test_sources[0], "category": categories[0], "value": values[0]},
    {"source": test_sources[0], "category": categories[0], "value": values[0]}, # Duplicate
    {"source": test_sources[0], "category": categories[1], "value": values[1]}, # Multiple categories for T1
    {"source": test_sources[1], "category": categories[2], "value": values[2]},
    {"source": test_sources[2], "category": categories[3], "value": values[3]},
    {"source": test_sources[3], "category": categories[4], "value": values[4]},
    {"source": test_sources[4], "category": categories[0], "value": values[4]},
    {"source": test_sources[4], "category": categories[1], "value": values[0]}  # Multiple categories for T5
]

# 2. Organized Test Data
organized_data = defaultdict(lambda: defaultdict(list))
seen_records = set()
duplicates = []

for rec in raw_records:
    rec_tuple = (rec["source"], rec["category"], rec["value"])
    if rec_tuple in seen_records:
        duplicates.append(rec)
    else:
        seen_records.add(rec_tuple)
        organized_data[rec["source"]][rec["category"]].append(rec["value"])

# 3. Numerical & Categorical Summaries
all_values = [rec["value"] for rec in raw_records]
distinct_values = sorted(list(set(all_values)))

print("=== 1. GENERATED TEST DATA ===")
for r in raw_records:
    print(r)

print("\n=== 2. REPEATED AND DISTINCT DATA ===")
print("Repeated Records:", duplicates)
print("Distinct Values:", distinct_values)

print("\n=== 3. ORGANIZED AND SUMMARY RESULTS ===")
for src, cat_dict in sorted(organized_data.items()):
    print(f"Source: {src}")
    for cat, vals in cat_dict.items():
        avg_val = sum(vals) / len(vals)
        print(f"  Category: {cat:<12} | Values: {vals} | Avg: {avg_val:.2f}")

print("\n=== 4. FINAL CONSOLIDATED SUMMARY ===")
print(f"Total Raw Records: {len(raw_records)}")
print(f"Total Unique Records: {len(seen_records)}")
print(f"Overall Dataset Average Value: {sum(all_values)/len(all_values):.2f}")
```

```python
# 2. Student Performance Data Analyzer
LAST_NAME = "Adonis"
SEED_NUM = 8
FAVORITE_ARTIST = "Gilbert Campo"

student_names = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

artist_clean = FAVORITE_ARTIST.replace(" ", "")
base_scores = [(SEED_NUM * 7 + ord(c.upper()) % 30) for c in artist_clean[:5]]

# 1. Generated Performance Records (Includes duplicate and incomplete record)
performance_records = [
    {"student": student_names[0], "subject": subjects[0], "score": base_scores[0]},
    {"student": student_names[0], "subject": subjects[0], "score": base_scores[0]}, # Duplicate record
    {"student": student_names[1], "subject": subjects[1], "score": base_scores[1]},
    {"student": student_names[2], "subject": subjects[2], "score": None},            # Incomplete record
    {"student": student_names[3], "subject": subjects[3], "score": base_scores[3]},
    {"student": student_names[4], "subject": subjects[4], "score": base_scores[4]},
]

def classify_performance(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 85:
        return "EXCELLENT"
    elif score >= 75:
        return "GOOD"
    else:
        return "NEEDS IMPROVEMENT"

# 2. Process Records
processed_records = []
seen_entries = set()
duplicates = []
incomplete = []

for rec in performance_records:
    entry_id = (rec["student"], rec["subject"], rec["score"])
    
    if rec["score"] is None:
        incomplete.append(rec)
    
    if entry_id in seen_entries:
        duplicates.append(rec)
    else:
        seen_entries.add(entry_id)
        category = classify_performance(rec["score"])
        processed_records.append({**rec, "category": category})

print("=== 1. GENERATED PERFORMANCE RECORDS ===")
for r in performance_records:
    print(r)

print("\n=== 2. REPEATED / INCOMPLETE RECORDS ===")
print("Repeated Records:", duplicates)
print("Incomplete Records:", incomplete)

print("\n=== 3. PROCESSED & ORDERED RESULTS ===")
valid_scores = [r["score"] for r in processed_records if r["score"] is not None]
for r in sorted(processed_records, key=lambda x: x["student"]):
    print(f"Student: {r['student']} | Subject: {r['subject']:<12} | Score: {str(r['score']):<4} | Category: {r['category']}")

print("\n=== 4. FINAL PERFORMANCE REPORT SUMMARY ===")
print(f"Total Entries: {len(performance_records)}")
print(f"Valid Evaluated Records: {len(valid_scores)}")
print(f"Class Average Score: {sum(valid_scores)/len(valid_scores):.2f}")
```

```python
# 3. Transaction History Analyzer
from collections import defaultdict

LAST_NAME = "Adonis"
SEED_NUM = 8
FAVORITE_ARTIST = "Gilbert Campo"

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {"NOTEBOOK": 50.0, "PEN": 20.0, "USB": 300.0, "HEADPHONES": 500.0, "CALCULATOR": 450.0}

artist_clean = FAVORITE_ARTIST.replace(" ", "")
quantities = [(SEED_NUM % 3 + 1 + (i % 3)) for i in range(5)]

# 1. Generate Transactions (Includes repeated customer-product purchase)
transactions = [
    {"customer": customers[0], "product": products[0], "qty": quantities[0]},
    {"customer": customers[0], "product": products[0], "qty": quantities[0]}, # Repeated purchase
    {"customer": customers[1], "product": products[1], "qty": quantities[1]},
    {"customer": customers[2], "product": products[2], "qty": quantities[2]},
    {"customer": customers[3], "product": products[3], "qty": quantities[3]},
    {"customer": customers[4], "product": products[4], "qty": quantities[4]},
    {"customer": customers[0], "product": products[2], "qty": quantities[1]}, # Additional purchase
]

# 2. Process and Calculate Totals
processed_txs = []
seen_txs = set()
duplicates = []

customer_totals = defaultdict(float)
product_totals = defaultdict(float)

for tx in transactions:
    item_total = tx["qty"] * prices[tx["product"]]
    tx_summary = {**tx, "unit_price": prices[tx["product"]], "total_value": item_total}
    
    tx_tuple = (tx["customer"], tx["product"], tx["qty"])
    if tx_tuple in seen_txs:
        duplicates.append(tx_summary)
    else:
        seen_txs.add(tx_tuple)
        processed_txs.append(tx_summary)
        customer_totals[tx["customer"]] += item_total
        product_totals[tx["product"]] += item_total

print("=== 1. GENERATED TRANSACTION RECORDS ===")
for tx in transactions:
    print(tx)

print("\n=== 2. REPEATED TRANSACTIONS ===")
print("Repeated Purchases:", duplicates)

print("\n=== 3. PROCESSED TRANSACTION VALUES ===")
for tx in processed_txs:
    print(f"Customer: {tx['customer']} | Item: {tx['product']:<10} | Qty: {tx['qty']} | Unit: ₱{tx['unit_price']} | Total: ₱{tx['total_value']}")

print("\n=== 4. CUSTOMER & PRODUCT SUMMARIES ===")
print("Customer Total Spending:")
for cust, total in customer_totals.items():
    print(f"  {cust}: ₱{total:.2f}")

print("\nProduct Revenue:")
for prod, total in product_totals.items():
    print(f"  {prod:<10}: ₱{total:.2f}")

print("\n=== 5. FINAL CONSOLIDATED TRANSACTION REPORT ===")
grand_total = sum(tx["total_value"] for tx in processed_txs)
print(f"Total Unique Transactions: {len(processed_txs)}")
print(f"Overall Dataset Transaction Value: ₱{grand_total:.2f}")
```
